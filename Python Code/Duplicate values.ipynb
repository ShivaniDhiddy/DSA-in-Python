### Linear Solution with linear time complexity

```python
def has_duplicate_value(array):
    # Using a set for O(1) average-time complexity lookups
    existing_numbers = set()
    
    for num in array:
        if num in existing_numbers:
            return True
        else:
            existing_numbers.add(num)
            
    return False
# this is faster because we are not storing any value 
```

### making a small chnage by using dictionaries instead of sets

```python
def has_duplicate_value(array):
    # Initialize an empty Python dictionary (hash map)
    existing_values = {}
    
    for item in array:
        # Check if the item is not already a key in the dictionary
        if item not in existing_values:
            existing_values[item] = 1
        else:
            return True
            
    return False
```
