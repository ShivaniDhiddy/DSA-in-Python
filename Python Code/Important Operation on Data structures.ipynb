## Arrays

- Arrays are most basic Data Structures, essential a list of elements or items stored in a contiguous memory locations.
- There are mainly 4 Array operations:
 - Read
 - Search
 - Insert
 - Delete
 
- Arrays are homogeous in nature unlike lists and require typecode at initialisation. examples are mentioned below:
![Source: Geeksforgeeks.com](Typecode.png)


```python
# Read => takes 1 step because machine can locate the element at index number 
# it can directly jump to the memory address of the array and add the 
# index number to find its location

import array as arr

my_array = arr.array('i',list(range(1,9))) #'i' is the typecode for intergers
```

```python
type(my_array)
```

```python
print(my_array) # if you print an array normally it will print the typecode as well
```

```python
print(*my_array) #using * unpacks the array elements and outputs just the elements.
```

### Insert elements into an array

```python
my_array.insert(1, 4) # insert element 4 at index 1
```

```python
print(*my_array)
```

### Accessing or reading elements 


```python
my_array[2:8]
```

### Search

```python
my_array.index(4)
```

```python
#Delete

my_array.remove(4)
```
